#QUESTION 8) Make two simple charts and write 3 observations

```python
age_arr = np.array(df["Age"])
np.mean(age_arr), np.median(age_arr), np.min(age_arr), np.max(age_arr)

xaxis = df["Age"]
plt.hist(xaxis)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()

group_nodes = df.groupby("Status")["Nodes"].mean()
plt.bar(group_nodes.index.astype(str), group_nodes.values)
plt.xlabel("Status")
plt.ylabel("Nodes")
plt.title("Bar Chart")
plt.show()
```
